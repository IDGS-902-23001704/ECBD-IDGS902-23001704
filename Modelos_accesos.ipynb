```python
# =========================================================
# MODELOS DE MACHINE LEARNING: SISTEMA DE ACCESOS IOT
# =========================================================

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, classification_report, mean_absolute_error, r2_score
```

```python
# ---------------------------------------------------------
# 1. MODELO DE CLASIFICACIÓN (Decidir si se abre la puerta)
# ---------------------------------------------------------
print("--- ENTRENANDO MODELO DE CLASIFICACIÓN (ACCESO) ---")

# Simulamos datos: [nivel_confianza_camara, hora_dia, residente_al_corriente_pago]
# Target: 1 (Acceso Permitido), 0 (Acceso Denegado)
X_clas = np.array([
    [0.98, 8, 1], [0.40, 14, 1], [0.95, 18, 0], [0.20, 3, 0],
    [0.99, 7, 1], [0.88, 19, 1], [0.35, 12, 1], [0.92, 22, 1]
])
y_clas = np.array([1, 0, 0, 0, 1, 1, 0, 1])
```

```python
# Dividir datos
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_clas, y_clas, test_size=0.25, random_state=42)
```

```python
# Entrenar Random Forest
clf = RandomForestClassifier(max_depth=5, random_state=42)
clf.fit(X_train_c, y_train_c)

# Evaluar
y_pred_c = clf.predict(X_test_c)
print(f"Exactitud: {accuracy_score(y_test_c, y_pred_c) * 100}%")
print("Reporte de Clasificación:\n", classification_report(y_test_c, y_pred_c, zero_division=0))
```

```python
# ---------------------------------------------------------
# 2. MODELO DE REGRESIÓN (Predecir tiempo de espera en fila)
# ---------------------------------------------------------
print("\n--- ENTRENANDO MODELO DE REGRESIÓN (TIEMPO DE ESPERA) ---")

# Simulamos datos: [hora_dia, vehiculos_en_fila]
# Target: Tiempo de espera estimado en segundos
X_reg = np.array([
    [8, 5], [14, 2], [18, 8], [3, 0],
    [7, 4], [19, 7], [12, 3], [22, 1]
])
y_reg = np.array([120, 45, 200, 10, 95, 180, 70, 25]) # Segundos de espera
```

```python
# Dividir datos
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_reg, y_reg, test_size=0.25, random_state=42)

# Entrenar Regresión Lineal
reg = LinearRegression()
reg.fit(X_train_r, y_train_r)

# Evaluar
y_pred_r = reg.predict(X_test_r)
print(f"Error Absoluto Medio (MAE): {mean_absolute_error(y_test_r, y_pred_r):.2f} segundos")
print(f"R2 Score: {r2_score(y_test_r, y_pred_r):.2f}")
```

```python
# Prueba predictiva
prueba_hora = 18
prueba_fila = 6
tiempo_estimado = reg.predict([[prueba_hora, prueba_fila]])
print(f"-> Predicción: A las {prueba_hora}:00 hrs con {prueba_fila} autos, la espera será de {tiempo_estimado[0]:.2f} segundos.")
```
